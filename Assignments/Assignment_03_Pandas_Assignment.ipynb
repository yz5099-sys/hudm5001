#### Exercise B1 — A Series of scores

1. Print the **average** score, rounded to 1 decimal.

**The setting.** Every analysis starts the same way: load the data and get your bearings. The **wine** dataset holds 178 wines from **three cultivars** (grape varieties), labeled `class_id` 1, 2, or 3, with thirteen chemical measurements each. You already set `WINE_URL` in the setup cell.

1. Read the CSV with `pd.read_csv(WINE_URL)` into a DataFrame called `wine`.
2. Print its **shape** and its **column names** as a list.

#### Exercise C2 — Filtering — find the strong, bold wines
**Difficulty:** ★★★★☆  **Points:** 12
**Skills:** new column, boolean filtering with `&`, `.shape`, arithmetic

**The setting.** A sommelier wants the full-bodied wines: high alcohol and deep color. This is boolean filtering — the same skill as NumPy masking, now on a real table. *(Uses the `wine` DataFrame from C1.)*

1. Add a `True`/`False` column `is_strong` that is `True` when `alcohol` is above **13**.
2. Filter for the wines that are **both** strong (`alcohol > 13`) **and** deeply colored (`color_intensity > 5`). Print how many there are with `.shape[0]`.
3. What **fraction** of all the wines are strong? Divide `wine["is_strong"].sum()` by the total number of rows (`wine.shape[0]`), rounded to 2 decimals.

````python
# Your code here
wine=pd.read_csv(WINE_URL)
wine['is_strong']=wine['alcohol']>13
print(wine[(wine.alcohol>13)&(wine.color_intensity>5)])
print(wine[(wine.alcohol>13)&(wine.color_intensity>5)].shape[0])
fraction=wine["is_strong"].sum()/wine.shape[0]
print(fraction.round(2))
````

#### Exercise C3 — Grouping — do the cultivars differ?
**Difficulty:** ★★★★☆  **Points:** 13
**Skills:** `groupby`, `.mean()`, `sort_values`, `value_counts`

**The setting.** The real question behind this dataset: are the three cultivars chemically different? `groupby` answers it in a couple of lines — split by cultivar, average each measurement, compare. *(Uses the `wine` DataFrame from C1.)*

1. Group by `class_id` and print the **mean** of `alcohol`, `flavanoids`, and `color_intensity` for each cultivar (select those three columns, then `.mean()`, rounded to 2).
2. Which cultivar has the highest average `proline`? Group `proline` by `class_id`, take the mean, and `sort_values(ascending=False)` so the winner is on top.
3. Print how many wines are in each cultivar with `value_counts()`.

````python
# Your code here
wine=pd.read_csv(WINE_URL)
print(wine.groupby('class_id')[['alcohol','flavanoids','color_intensity']].mean().round(2))
print(wine.groupby('class_id')['proline'].mean().sort_values(ascending=False).head(1))
print(wine['class_id'].value_counts())
````

#### Exercise C4 — Capstone — a small end-to-end analysis
**Difficulty:** ★★★★★  **Points:** 13
**Skills:** filtering, new column, `groupby`, `set()`, `json.dump`/`json.load`, `os.path.exists`, f-string

**The setting.** Put it all together the way a real analysis flows: subset, summarize, then **save the results** so a later step (or a teammate) can pick them up. This pulls in the file I/O from Session 02 as well. *(Uses the `wine` DataFrame from C1.)*

1. Add a `True`/`False` column `high_proline` that is `True` when `proline` is above the **mean** `proline`. Print how many wines are high-proline (`.sum()`).
2. Pull out just the **cultivar-1** wines (`class_id == 1`) and print their **average alcohol**, rounded to 2.
3. Print the **mean alcohol per cultivar** with `groupby`.
4. Count the distinct cultivars with `len(set(wine["class_id"]))`. Build a small **summary dictionary** with the number of wines, the number of cultivars, the mean alcohol, and the count of high-proline wines. **Write it** to `wine_summary.json` with `json.dump`, read it back with `json.load`, and confirm the file exists with `os.path.exists`.
5. Finish with an f-string, for example: `The wine set has 178 wines across 3 cultivars, averaging 13.0% alcohol.`

````python
# Your code here
wine=pd.read_csv(WINE_URL)
wine['high_proline']=wine.proline>wine.proline.mean()
print(wine['high_proline'].sum())
print(wine[wine['class_id']==1]['alcohol'].mean().round(2))
print(wine.groupby('class_id')['alcohol'].mean())
print(len(set(wine["class_id"])))
dic={
    'number_wines':wine.shape[0],
    'number_cultivars':len(set(wine["class_id"])),
    'mean_alcohol':wine['alcohol'].mean(),
    'high_proline_wines':int(wine['high_proline'].sum())
}
with open('wine_summary.json','w') as file:
  json.dump(dic,file)
with open('wine_summary.json','r') as file:
  content=json.load(file)
print(content)
print(os.path.exists('wine_summary.json'))
print(
    f"The wine set has {dic['number_wines']} wines "
    f"across {dic['number_cultivars']} cultivars, "
    f"averaging {dic['mean_alcohol']:.1f}% alcohol."
)
````
