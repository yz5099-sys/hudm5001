#### Exercise C2 — NLP — clean a review, then save the stats
**Difficulty:** ★★★★★  **Points:** 12
**Skills:** `.lower()`, `.replace()`, `.split()`, `set()`, `len()`, arithmetic, `json.dump` / `json.load`, f-string

**The setting.** A text-analysis pipeline usually *ends* by saving its results so a later step can pick them up. You will clean a short review (Session 01 string work) and write the numbers out to a JSON file (Session 02 I/O).

```python
review = "The food was good and the service was good and the price was fair"
```

Treat `"the"`, `"was"`, and `"and"` as stop words.

1. **Lowercase** the review, split the text into a list of words, count the number of words.
2. Count the **unique** words by converting the word list to a `set()`.
3. Compute a rough **lexical diversity** score, `number of unique words / number of total words`, rounded to 2 decimals. Print the result using f-string.
4. Remove the stop words with the padding trick from class: pad the lowercased text with a leading and trailing space, then chain `.replace()` functions. Split the result and count what is left.
5. You may wonder why we should remove the stop words when analyzing students' writting samples. Now, please re-compute the **lexical diversity** using the cleaned text. Print the result using f-sting.
6. Put number of total and unqiue words on both the uncleaned and the cleaned texts, the lexical diversity scores for both uncleaned and the cleaned text in a dictionary, **write it to `review_stats.json`** with `json.dump`, then read it back with `json.load` and print it.
7. Finish with an f-string: `___ words, ___ unique, ___ after cleaning. Diversity = ___.`  

**Tips:** Although this is still a toy example, it might be your longest Python script so far (over 40 lines probably). You will experience a complete workflow in computational linguistic: load text --> clean text --> calculate (model fit) --> save the results. Keep calm and code! 😎

````python
# Your code here
review = "The food was good and the service was good and the price was fair"
review=review.lower()
review_split=review.split()
total_words1=len(review_split)
print(total_words1)
review_unique=set(review_split)
unique_words1=len(review_unique)
print(unique_words1)
Lexical_diversity1=round(unique_words1 / total_words1,2)
print(f'the lexical diversity is {unique_words1} / {total_words1}, which is {Lexical_diversity1}')

review=' '+review+' '
review=review.replace('the',' ')
review=review.replace('and',' ')
review=review.replace('was',' ')
review=review.split()
left_words=len(review)
print(left_words)
total_words2=left_words
print(total_words2)
review_unique=set(review)
unique_words2=len(review_unique)
print(unique_words2)
Lexical_diversity2=round(unique_words2 / total_words2,2)
print(f'the lexical diversity is {unique_words2} / {total_words2}, which is {Lexical_diversity2}')

dic={'uncleaned':{'totalwords':total_words1,'uniquewords':unique_words1,'Lexicaldiversity':Lexical_diversity1},'cleaned':{'totalwords':total_words2,'uniquewords':unique_words2,'Lexicaldiversity':Lexical_diversity2}}
with open('review_stats.json','w') as file:
  json.dump(dic,file)
with open('review_stats.json','r') as file:
  back=json.load(file)
print(back)
print(f'{total_words1} words, {unique_words1} unique, {total_words2} after cleaning. Diversity = {Lexical_diversity2}.')
````

#### Exercise C4 — Real data — a first look at the iris dataset
**Difficulty:** ★★★★★  **Points:** 13
**Skills:** `pd.read_csv` (from a URL), `set()`, `np.array`, `.mean()` / `.max()` / `.min()`, boolean indexing, `%`, `json.dump` / `json.load`, `os.path.exists`, f-string

**The setting.** This is the whole toolkit at once — read a real dataset from the web, poke at it with NumPy, and save a small summary. That is the shape of an actual analysis. The famous **iris** dataset holds 150 flowers from three species, with four measurements each. You already set `IRIS_URL` in the setup cell.

1. Read the CSV with `pd.read_csv(IRIS_URL)` into `iris`. Print its **shape** and its **column names** as a list.
2. The `species` column repeats a lot. Wrap it in `set(iris["species"])` to get the **unique species**, and print how many there are with `len()`.
   *(Selecting a column by name, `iris["column"]`, works just like a dictionary lookup from Session 01.)*
3. Turn the `petal_length` column into a NumPy array: `petals = np.array(iris["petal_length"])`. Print its **mean** (rounded to 2), **max**, and **min**.
4. Using **boolean indexing**, count how many flowers have a petal length **above the mean**: `(petals > petals.mean()).sum()`.
5. Use the **modulus** to print whether the number of flowers is even: `len(petals) % 2 == 0`.
6. Build a small summary **dictionary** (total number of flowers, number of species, mean petal length, how many above the mean), write it to `iris_summary.json` with `json.dump`, read it back with `json.load`, and confirm it is there with `os.path.exists`.
7. End with an f-string: `iris has ___ flowers across ___ species; ___ have longer-than-average petals.`

````python
# Your code here
iris=pd.read_csv(IRIS_URL)
print(iris.shape)
print(list(iris.columns))
count_species=len(set(iris["species"]))
print(count_species)
petals = np.array(iris["petal_length"])
print(petals.mean().round(2),petals.max(),petals.min())
print((petals > petals.mean()).sum())
print(len(petals) % 2 == 0)
total_flowers=len(petals)
above_mean=int((petals > petals.mean()).sum())
summary = {
    "total_flowers": total_flowers,
    "number_of_species": count_species,
    "mean_petal_length": float(petals.mean().round(2)),
    "above_mean": above_mean
}
with open ('iris_summary.json','w') as file:
  json.dump(summary,file)
with open('iris_summary.json','r') as file:
  back=json.load(file)
print(back)
print(os.path.exists('iris_summary.json'))
print(f'iris has {total_flowers} flowers across {count_species} species; {above_mean} have longer-than-average petals.')
````
